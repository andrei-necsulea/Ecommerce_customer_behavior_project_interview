```python
import pandas as pd

df = pd.read_csv("feature_engineering.csv")
df1 = pd.read_csv("ecommerce_customer_behavior_dataset.csv")

#Gasim coloanele care sunt în df1 dar nu in df
cols_to_add = [col for col in df1.columns if col not in df.columns]

print(f"Coloane care lipsesc din df: {cols_to_add}")

#Adaugam aceste coloane la df (aliniate pe randuri)
for col in cols_to_add:
    df[col] = df1[col]

#Verificare
print(f"Noul df are acum {df.shape[1]} coloane (față de {len(df.columns) - len(cols_to_add)} anterior).")
print(f"Coloanele adăugate: {cols_to_add}")

df.to_csv("feature_engineering_extended.csv", index=False, encoding='utf-8-sig')
```
