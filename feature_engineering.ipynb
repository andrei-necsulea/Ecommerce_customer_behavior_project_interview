```python
import pandas as pd

pd.set_option('display.max_columns', None)
```

```python
df = pd.read_csv("ecommerce_customer_behavior_dataset.csv")
```

```python
df.head()
```

```python
id_list = df['Order_ID'].tolist()

for i in range(len(id_list)):
    id_list[i] = id_list[i][4:]
    for j in range(len(id_list[i])):
        if id_list[i][j] == 0:
            id_list[i][j] = id_list[i][j].replace("0","")
    id_list[i] = int(id_list[i])
```

```python
id_list
```

```python
df['Customer_ID'] = id_list
df.drop('Order_ID', axis=1, inplace=True)
```

```python
df.head()
```

```python
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Day'] = df['Date'].dt.day
df['DayOfWeek'] = df['Date'].dt.dayofweek
df['Quarter'] = df['Date'].dt.quarter
df['IsWeekend'] = df['DayOfWeek'].isin([5, 6]).astype(int)
df = df.drop(columns='Date')
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df['Discount_Percentage'] = (df['Discount_Amount'] / (df['Unit_Price'] * df['Quantity'])) * 100
```

```python
df['Avg_Spending_City'] = df.groupby('City')['Total_Amount'].transform('mean')
```

```python
total_amount = df["Total_Amount"]
df.drop(columns=['Discount_Amount', 'Total_Amount'], inplace=True)
```

```python
df.head()
```

```python
df['City'].unique()
```

```python
df.drop(columns='City', inplace=True)
```

```python
df.head()
```

```python
#df['Gender'] = df['Gender'].map({'Male' : '1', 'Female' : '0'})
```

```python
#df.head()
```

```python
df['Product_Category'].unique()
```

```python
df = pd.get_dummies(df, columns=['Product_Category'], prefix='Product')
```

```python
df.head()
```

```python
df['Payment_Method'].unique()
```

```python
payment_method = df['Payment_Method']
df = pd.get_dummies(df, columns=['Payment_Method'], prefix='Payment')
```

```python
df.head()
```

```python
df['Device_Type'].unique()
```

```python
df = pd.get_dummies(df, columns=['Device_Type'], prefix='Device')
```

```python
df.head()
```

```python
len(df['Customer_ID'].unique())
```

```python
df.drop(columns='Customer_ID', inplace=True)
```

```python
df.head()
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
df['Engagement_Score'] = scaler.fit_transform(
    (df['Session_Duration_Minutes'] * df['Pages_Viewed']).values.reshape(-1,1)
)

df['Pages_per_Minute'] = df['Pages_Viewed'] / df['Session_Duration_Minutes']

df['Browsing_to_Delivery_Ratio'] = df['Pages_per_Minute'] / df['Delivery_Time_Days']

df.drop(columns=['Session_Duration_Minutes', 'Pages_Viewed'], inplace=True)
```

```python
df.head()
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df['Customer_Rating'] = scaler.fit_transform(df[['Customer_Rating']])
```

```python
df.head()
```

```python
avg_rating = df['Customer_Rating'].mean()
df['Rating_Deviation'] = df['Customer_Rating'] - avg_rating

df.drop(columns=['Customer_Rating'], inplace=True)
```

```python
df.head()
```

```python
bins = [0, 18, 25, 35, 50, 65, 100]
labels = ['Teen', 'Young_Adult', 'Adult', 'Middle_Aged', 'Senior', 'Elder']

df['Age_Group'] = pd.cut(df['Age'], bins=bins, labels=labels)
age_group = df['Age_Group']

df['Demo_Segment'] = df['Gender'].astype(str) + '_' + df['Age_Group'].astype(str)
demo_seg = df['Demo_Segment']
```

```python
df = pd.get_dummies(df, columns=['Age_Group'], prefix='Age')
df = pd.get_dummies(df, columns=['Demo_Segment'], prefix='Seg')
```

```python
df.head()
```

```python
df['Total_Amount'] = total_amount
df['Avg_Spending_Gender'] = df.groupby('Gender')['Total_Amount'].transform('mean')
df['Relative_Spending'] = df['Total_Amount'] / df['Avg_Spending_Gender']
```

```python
df['Age_Group'] = age_group
df['Avg_Spending_AgeGroup'] = df.groupby('Age_Group')['Total_Amount'].transform('mean')
df['AgeGroup_Spending_Deviation'] = df['Total_Amount'] - df['Avg_Spending_AgeGroup']
```

```python
df['Engagement_per_Year'] = df['Engagement_Score'] / df['Age']
```

```python
df['Payment_Method'] = payment_method
df['Gender_Payment'] = df['Gender'] + '_' + df['Payment_Method']
df = pd.get_dummies(df, columns=['Gender_Payment'])
```

```python
df['Gender'] = df['Gender'].map({'Male' : 1, 'Female' : 0, 'Other' : -1})

df['Age_x_Gender'] = df['Age'] * df['Gender']
```
