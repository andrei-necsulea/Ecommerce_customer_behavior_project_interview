```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegressionCV

lr = LogisticRegressionCV(max_iter=100)
lr.fit(X, y)

coefs = pd.Series(lr.coef_[0], index=X.columns)
coefs.abs().sort_values(ascending=False).head(76).plot(kind='barh', title='Logistic Coefficients', figsize=(12, 20))
```

```python
low_importance = coefs[coefs.abs() < 0.25].index.tolist()
```

```python
low_importance
```

```python
# coeficienți (Series: index = nume coloane)
coefs = pd.Series(lr.coef_[0], index=X.columns)

# prag
thr = 0.25

# lista inițială (|coef| < thr)
low_importance = coefs[coefs.abs() < thr].index.tolist()

# 1) coloane object (dacă mai există)
object_cols = X.select_dtypes(include=['object']).columns

# 2) alternativ: încearcă să detectezi dummies (0/1)
if len(object_cols) == 0:
    dummy_cols = [c for c in X.columns
                  if set(pd.unique(X[c].dropna())) <= {0,1}]
    categorical_like = dummy_cols
else:
    categorical_like = object_cols

# filtrează coeficienții mici din „categorical_like”
filtered = coefs[(coefs.abs() < thr) & (coefs.index.isin(categorical_like))]

if filtered.empty:
    print("⚠️ Nu există nicio coloană cu |coef| < 0.25 care să fie object/dummy.")
else:
    # cea mai mare (în valoare absolută) dintre ele
    max_feature = filtered.abs().idxmax()
    max_value = coefs[max_feature]
    print(f"📈 Păstrez: {max_feature} (coef={max_value:.4f})")

    # scoate-o din low_importance (dacă e acolo)
    if max_feature in low_importance:
        low_importance.remove(max_feature)
        print(f"✅ '{max_feature}' a fost eliminat din lista low_importance.")
    else:
        print(f"ℹ️ '{max_feature}' nu era în low_importance.")
```

```python
from sklearn.base import ClassifierMixin, RegressorMixin
```

```python
results = {}

for name, est in models.items():
    est.fit(X_train, y_train)
    if isinstance(est, ClassifierMixin):
        y_pred = est.predict(X_test)
        y_score = est.predict_proba(X_test)[:,1] if hasattr(est,"predict_proba") else (
                  est.decision_function(X_test) if hasattr(est,"decision_function") else None)
    elif isinstance(est, RegressorMixin):
        y_score = est.predict(X_test)
        y_pred  = (y_score >= 0.5).astype(int)
    else:
        continue

    auc = roc_auc_score(y_test, y_score, multi_class="ovr") if y_score is not None else float("nan")
    f1  = f1_score(y_test, y_pred)
    acc = (y_pred == y_test).mean()

    print(f"\n=== {name} ===")
    print(f"AUC={auc:.3f} | F1={f1:.3f} | Acc={acc:.3f}")
    print("Confusion:\n", confusion_matrix(y_test, y_pred))
    print(classification_report(y_test, y_pred, digits=3))

    results[name] = {"model": est, "auc": auc, "f1": f1, "acc": acc}

winner = max(results.items(), key=lambda kv: kv[1]["f1"])
print(f"\nBest model by F1: {winner[0]}  (F1={winner[1]['f1']:.3f} , acc={winner[1]['acc']*100:.2f}%)")
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix, classification_report, roc_curve
import numpy as np

def evaluate_model(model, X_test, y_test):
    """
    Afișează metricele clasice de evaluare pentru un model de clasificare binară.
    """

    # 🔹 Predicții
    y_pred = model.predict(X_test)

    # 🔹 Probabilități (pentru ROC-AUC)
    if hasattr(model, "predict_proba"):
        y_proba = model.predict_proba(X_test)[:, 1]
    elif hasattr(model, "decision_function"):
        y_proba = model.decision_function(X_test)
    else:
        y_proba = np.zeros_like(y_pred, dtype=float)

    # 🔹 Calcul metrici
    acc  = accuracy_score(y_test, y_pred)
    prec = precision_score(y_test, y_pred, zero_division=0)
    rec  = recall_score(y_test, y_pred)
    f1   = f1_score(y_test, y_pred)
    auc  = roc_auc_score(y_test, y_proba)

    # 🔹 Printare frumoasă
    print(" MODEL EVALUATION RESULTS")
    print("="*40)
    print(f" Accuracy:   {acc:.4f}")
    print(f" Precision:  {prec:.4f}")
    print(f" Recall:     {rec:.4f}")
    print(f" F1-Score:   {f1:.4f}")
    print(f" ROC-AUC:    {auc:.4f}")
    print("="*40)
    print("\nConfusion Matrix:")
    print(confusion_matrix(y_test, y_pred))
    print("\nClassification Report:")
    print(classification_report(y_test, y_pred, digits=3))

    return {
        "accuracy": acc,
        "precision": prec,
        "recall": rec,
        "f1": f1,
        "auc": auc
    }
```

```python
evaluate_model(lr, X_test, y_test)
```
